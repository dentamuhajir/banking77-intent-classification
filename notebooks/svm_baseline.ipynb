# Baseline TF-IDF–SVM Banking77
Gunakan runtime CPU dan jalankan semua sel. Pilih akun yang memiliki folder `Skripsi Banking77` ketika diminta memasang Drive. Model dan laporan validasi akan disimpan ke folder proyek tersebut.

```python
from pathlib import Path
import subprocess
import sys

from google.colab import drive

drive.mount("/content/drive")
project = Path("/content/drive/MyDrive/Skripsi Banking77")
runner = project / "notebooks" / "runner"
subprocess.run([sys.executable, "-m", "pip", "install", "-r", str(runner / "requirements.txt")], check=True)
```

```python
subprocess.run([
    sys.executable, str(runner / "03_train_svm.py"),
    "--data-dir", str(project / "data"), "--output-dir", str(project),
    "--run-name", "colab",
], check=True)
```

```python
import json

report = json.loads((project / "results/svm_baseline_colab.json").read_text())
demo = subprocess.check_output([
    sys.executable, str(runner / "predict_cli.py"),
    "--model", str(project / "models" / report["model_filename"]), report["demo_text"],
], text=True).strip()
assert demo == report["demo_label"]
print(report["demo_text"], "->", demo)
print("SHA-256 model untuk pemeriksaan di laptop:", report["model_sha256"])
```
